```python
import pandas as pd
import numpy as np

data = pd.read_csv(r"data\sonar.csv")
data_arr = np.asarray(data)
features = data_arr[:, 0:60]
label = data_arr[:, -1]
data.describe()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(features, label, test_size = 0.33, random_state = 69)
```

```python
from sklearn.naive_bayes import GaussianNB
nb = GaussianNB()
nb.fit(X_train, y_train)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, ConfusionMatrixDisplay
y_pred_nb = nb.predict(X_test)
print("Accuracy: ", accuracy_score(y_test, y_pred_nb))
print('\n')
print(classification_report(y_test, y_pred_nb))
cm_plot = ConfusionMatrixDisplay(confusion_matrix(y_test, y_pred_nb))
cm_plot.plot()
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(X_train, y_train)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, ConfusionMatrixDisplay
y_pred_lr = lr.predict(X_test)
print("Accuracy: ", accuracy_score(y_test, y_pred_lr))
print('\n')
print(classification_report(y_test, y_pred_lr))
cm_plot_lr = ConfusionMatrixDisplay(confusion_matrix(y_test, y_pred_lr))
cm_plot_lr.plot()
```
